```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(16,10))
plt.xlabel('Date', fontsize=14)
plt.ylabel('Number of Posts', fontsize=14)
plt.ylim(0, 35000)

for column in reshaped_df.columns:
    plt.plot(reshaped_df.index, reshaped_df[column], label=reshaped_df[column].name)

plt.legend(fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(16,10))
plt.xlabel('Date',fontsize = 14)
plt.ylabel('Number of Posts', fontsize = 14)
plt.ylim(0, 35000)

for column in reshaped_df:
  plt.plot(reshaped_df.index, reshaped_df[column], label = reshaped_df[column].name)

plt.legend(fontsize = 16)
plt.show()
```

```python
rolling_df = reshaped_df.rolling(window=30).mean()

plt.figure(figsize=(16,10))
plt.xlabel('Date', fontsize=1)
plt.ylabel('Number of Posts', fontsize=14)
plt.ylim(0, 35000)

for column in rolling_df:
  plt.plot(rolling_df.index, rolling_df[column], label=rolling_df[column].name)

plt.legend(fontsize=16)
plt.show()
```
